```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D


# Parameters for the LCG
M = 244944  # Modulus
a = 1597  # Multiplier
b = 51749  # Increment

print(M)
print(f'We find that it is {a< M and b< M} that a,b < M')

# Initialize arrays
U = np.zeros(10**4, dtype=np.int64)  # Raw integer values
U_final = np.zeros(10**4, dtype=np.float64)  # Normalized [0, 1)

# Seed
U[0] = 12345

# LCG computation
for i in range(1, 10**4):
    U[i] = (U[i-1] * a + b) % M  # Compute the next raw value
    U_final[i] = U[i] / M  # Normalize to [0, 1)

plt.figure(1)
#plt.hist(U_final,bins=int(np.sqrt(len(U_final))),density=True)
plt.hist(U_final,bins=50,density=True)


plt.figure(2)
plt.scatter(U_final[0:-1],U_final[1:], s = 1)

fig = plt.figure(3)  # Create a new figure
ax = fig.add_subplot(111, projection='3d')  # Add a 3D subplot
ax.scatter(U_final[:-2], U_final[1:-1], U_final[2:], c='b', marker='o',s=1)

```
