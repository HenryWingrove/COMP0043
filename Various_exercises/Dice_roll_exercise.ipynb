```python
import numpy as np
from scipy.integrate import quad

n = 100
m = 340

def phi_x(xi):
    summer = 0
    for i in range(1,7):
        summer += np.exp(1j*i*xi)
    return (summer/6)
def phi_yn(xi):
    return (phi_x(xi)) ** n

def real_int(xi):
    return (1/(2*np.pi))*(np.exp(-1j*m*xi)*phi_yn(xi)).real

def imag_int(xi):
    return (1/(2*np.pi))*(np.exp(-1j*m*xi)*phi_yn(xi)).imag


real_int_val, _ = quad(real_int, -np.pi,np.pi)
imag_int_val, _ = quad(imag_int, -np.pi,np.pi)

P_Y_M = (real_int_val + 1j* imag_int_val)

print(P_Y_M)


```

```python
#Now we do a little montecarlo
import random as rand

trials = 10**6
hit = 0
for i in range(trials):
    summer = 0
    for i in range(100):
        summer += int(rand.random()*6) + 1 

    if summer == 340:
        hit+= 1
print(hit)
print(f'probability is {hit/trials}')
```
