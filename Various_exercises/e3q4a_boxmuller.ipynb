```python
import numpy as np
import matplotlib.pyplot as plt
import random as rand

# Box-Muller transform functions
def z0(U1, U2):
    return np.sqrt(-2 * np.log(U1)) * np.cos(2 * np.pi * U2)

def z1(U1, U2):
    return np.sqrt(-2 * np.log(U1)) * np.sin(2 * np.pi * U2)

# Number of samples
samples = 10**4

# Arrays to store results
z0_arr = np.zeros(samples)
z1_arr = np.zeros(samples)

# Generate samples
for i in range(samples):
    U1, U2 = rand.random(), rand.random()  # Uniform random numbers
    z0_arr[i] = z0(U1, U2)
    z1_arr[i] = z1(U1, U2)

# Plot histogram of z0
plt.figure(1)
plt.hist(z0_arr, bins=50, density=True)
plt.title('Histogram of z0 (Box-Muller)')
plt.xlabel('Value')
plt.ylabel('Density')

plt.show()
```
