```python
import numpy as np
import random as rand
import matplotlib.pyplot as plt


n = 10**6
simulated = np.empty(n)


for i in range(n):
    simulated[i] = np.cos((np.pi*rand.random())-(np.pi/2))

plt.figure(1)
#plt.hist(U_final,bins=int(np.sqrt(len(U_final))),density=True)
plt.hist(simulated,bins=50,density=True)


def analytic(y):
    return 2/(np.pi*np.sqrt(1-y**2))

nsteps = 2048
grid = np.linspace(0,1,nsteps+1)


plt.figure(1)
#plt.hist(U_final,bins=int(np.sqrt(len(U_final))),density=True)
plt.hist(simulated,bins=50,density=True)
plt.plot(grid,analytic(grid))

```
