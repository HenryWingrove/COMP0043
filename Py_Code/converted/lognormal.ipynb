```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import lognorm, norm

# Define the parameters
mu = 0.2  # Mean of the underlying normal distribution
sigma = 0.1  # Standard deviation of the underlying normal distribution
nsample = 10**6  # Number of random samples

# Define the grid
x = np.arange(0.02, 2.02, 0.02)  # Avoid x = 0 to prevent division by zero in lognormal PDF

# Compute the PDF and CDF
f = lognorm.pdf(x, s=sigma, scale=np.exp(mu))  # PDF of the lognormal distribution
F = lognorm.cdf(x, s=sigma, scale=np.exp(mu))  # CDF of the lognormal distribution

# Plot the PDF and CDF
plt.figure(1)
plt.plot(x, f, 'r', label='PDF')
plt.plot(x, F, 'b', label='CDF')
plt.xlabel('x')
plt.legend()
plt.title('Lognormal distribution with $\mu = 0.2$ and $\sigma = 0.1$')
plt.savefig('lognormal.pdf')
plt.show()

# Sample from the underlying normal distribution
X = mu + sigma * np.random.randn(nsample)  # Normal random variables
Y = np.exp(X)  # Lognormal random variables via exponential transformation

# Histogram and fit of lognormal distribution
plt.figure(2)
plt.hist(Y, bins=100, density=True, alpha=0.6, color='b', label='Sampled')
x_fit = np.linspace(0.02, 2, 100)
f_fit = lognorm.pdf(x_fit, s=sigma, scale=np.exp(mu))
plt.plot(x_fit, f_fit, 'r', label='Lognormal fit')
plt.legend()
plt.title('Lognormal random variables with $\mu = 0.2$ and $\sigma = 0.1$')
plt.show()

# Bin the random variables in a histogram
dx = x[1] - x[0]
h, edges = np.histogram(Y, bins=x, density=False)
bin_centers = (edges[:-1] + edges[1:]) / 2

# Plot histogram
plt.figure(3)
plt.plot(bin_centers, h, label='Binned samples')
plt.title('Lognormal random variables with $\mu = 0.2$ and $\sigma = 0.1$')
plt.xlabel('x')
plt.ylabel('Frequency')
plt.legend()
plt.show()

# Bar plot of histogram
plt.figure(4)
plt.bar(bin_centers, h, width=dx, color='b', alpha=0.6, label='Binned samples')
plt.title('Lognormal random variables with $\mu = 0.2$ and $\sigma = 0.1$')
plt.xlabel('x')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
