```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
N = 1024  # number of grid points
Dx = 0.1  # spacing of the x grid
x = Dx * np.arange(-N // 2, N // 2)  # grid in x space
L = N * Dx  # truncation limit in x space
Dxi = 2 * np.pi / L  # spacing of the xi grid; Nyquist relation: Dx*Dxi = 2*pi/N
xi = Dxi * np.arange(-N // 2, N // 2)  # grid in xi space
W = N * Dxi  # truncation limit in xi space; W = 2*pi/Dx

# Define the parameter `a`
a = 2

# Analytical functions
fa = np.sqrt(a / np.pi) * np.exp(-a * x**2)  # Gaussian in x space
Fa = np.exp(-xi**2 / (4 * a))  # Gaussian in xi space

# Numerical computation
Fn = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(fa))) * L
fn = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(Fa))) / L
Fn1 = np.fft.fftshift(np.fft.fft(np.fft.ifftshift(fa))) * Dx
fn1 = np.fft.fftshift(np.fft.ifft(np.fft.ifftshift(Fa))) / Dx

# Plot results
# Plot 1: Real and Imaginary parts of fn vs. Analytical fa
plt.figure(1)
plt.plot(x, np.real(fn), 'r', label='Re(fn)')
plt.plot(x, np.imag(fn), 'g', label='Im(fn)')
plt.plot(x, fa, 'y:', label='fa')
plt.axis([-5, 5, 0, 1])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Numerical vs Analytical (fn)')
plt.grid(True)
plt.show()

# Plot 2: Real and Imaginary parts of Fn vs. Analytical Fa
plt.figure(2)
plt.plot(xi, np.real(Fn), 'b', label='Re(Fn)')
plt.plot(xi, np.imag(Fn), 'm', label='Im(Fn)')
plt.plot(xi, Fa, 'c:', label='Fa')
plt.axis([-10, 10, 0, 1])
plt.xlabel('ξ')
plt.ylabel('F')
plt.legend()
plt.title('Numerical vs Analytical (Fn)')
plt.grid(True)
plt.show()

# Plot 3: Real and Imaginary parts of fn1 vs. Analytical fa
plt.figure(3)
plt.plot(x, np.real(fn1), 'r', label='Re(fn1)')
plt.plot(x, np.imag(fn1), 'g', label='Im(fn1)')
plt.plot(x, fa, 'y:', label='fa')
plt.axis([-5, 5, 0, 1])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Numerical vs Analytical (fn1)')
plt.grid(True)
plt.show()

# Plot 4: Real and Imaginary parts of Fn1 vs. Analytical Fa
plt.figure(4)
plt.plot(xi, np.real(Fn1), 'b', label='Re(Fn1)')
plt.plot(xi, np.imag(Fn1), 'm', label='Im(Fn1)')
plt.plot(xi, Fa, 'c:', label='Fa')
plt.axis([-10, 10, 0, 1])
plt.xlabel('ξ')
plt.ylabel('F')
plt.legend()
plt.title('Numerical vs Analytical (Fn1)')
plt.grid(True)
plt.show()
```
