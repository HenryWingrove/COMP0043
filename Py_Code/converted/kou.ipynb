```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
eta1 = 4
eta2 = 3
p = 0.4
xmax = 2  # truncation
deltax = 0.01  # grid step
binw = 0.1  # bin width
n = 10**6  # number of random samples

# Compute the PDF
x = np.arange(-xmax, xmax + deltax, deltax)  # grid
fX = (
    p * eta1 * np.exp(-eta1 * x) * (x >= 0)
    + (1 - p) * eta2 * np.exp(eta2 * x) * (x < 0)
)  # PDF

# Sample the distribution
U = np.random.rand(n)  # standard uniform random variable
X = (
    -1 / eta1 * np.log((1 - U) / p) * (U >= 1 - p)
    + 1 / eta2 * np.log(U / (1 - p)) * (U < 1 - p)
)  # bilateral exponential random variable

# Plot
plt.figure(figsize=(10, 6))
x2 = np.arange(-xmax, xmax + binw, binw)  # bin edges
plt.hist(X, bins=x2, density=True, alpha=0.5, label='Sampled')  # histogram of sampled data
plt.plot(x, fX, 'r-', linewidth=2, label='Theory')  # theoretical PDF
plt.xlabel('x')
plt.ylabel('$f_X(x)$')
plt.legend()
plt.title('Asymmetric Double-Sided Exponential Distribution')
plt.grid(True)
plt.show()
```
