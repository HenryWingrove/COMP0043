```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon

# Define the parameters
mu = 2 / 3  # mean
nsample = 10**6  # number of random samples

# Define the grid
x= np.linspace(0,5,101)

# Compute the PDF and the CDF
f = expon.pdf(x, scale=mu)  # PDF of exponential distribution
F = expon.cdf(x, scale=mu)  # CDF of exponential distribution


# Plot the PDF and CDF
plt.figure(1)
plt.plot(x, f, 'r', label='PDF')
plt.plot(x, F, 'b', label='CDF')
plt.xlabel('x')
plt.legend()
plt.title('Exponential distribution with $\mu = 2/3$')
plt.savefig('exponential.pdf')  # Save the figure as a PDF
plt.show()

# Sample an exponential distribution
U = np.random.rand(nsample)  # standard uniform random numbers
X = -mu * np.log(U)  # transformation to exponential random numbers

# Bin the random variables in a histogram and normalize it
dx = x[1] - x[0]  # bin width
h, xx = np.histogram(X, bins=x + dx / 2, density=True)
xx = xx[:-1]  # Use the bin centers for plotting

# Plot sampled histogram vs theoretical PDF
plt.figure(2)
plt.plot(xx, h, 'b', label='Sampled')
plt.plot(x, f, 'r', label='Theory')
plt.xlim([0, 5])
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Exponential distribution with $\mu = 2/3$')
plt.show()

# Bar plot of histogram with overlaid PDF
plt.figure(3)
plt.bar(xx, h, width=dx, color='b', alpha=0.5, label='Sampled')  # Bar histogram
plt.plot(x, f, 'r', label='Theory')  # Overlay theoretical PDF
plt.xlabel('x')
plt.ylabel('f')
plt.legend()
plt.title('Exponential distribution with $\mu = 2/3$')
plt.show()
```
