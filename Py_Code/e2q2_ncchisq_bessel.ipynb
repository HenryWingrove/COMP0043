```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ncx2
from scipy.special import iv  # Modified Bessel function of the first kind

# Parameters for the noncentral chi-squared distribution
k = 5  # Degrees of freedom
lambda_ = 2  # Noncentrality parameter
x = np.linspace(0, 20, 500)  # Range of x values

# PDF using the library function
pdf_library = ncx2.pdf(x, df=k, nc=lambda_)

# PDF using the formula with the modified Bessel function
pdf_bessel = (0.5 * np.exp(-(x + lambda_) / 2)
              * ((x / lambda_) ** ((k / 4) - 0.5))
              * iv((k / 2) - 1, np.sqrt(lambda_ * x)))

# Plotting both PDFs for comparison
plt.figure(figsize=(8, 6))
plt.plot(x, pdf_library, label="Library Function (scipy.stats.ncx2)", color="blue")
plt.plot(x, pdf_bessel, label="Bessel Function Formula", linestyle="dashed", color="red")
plt.xlabel("x")
plt.ylabel("PDF")
plt.title("Noncentral Chi-Squared PDF Comparison")
plt.legend()
plt.grid()
plt.show()
```
